```python
import pandas as pd
import networkx as nx
import math
import heapq
import time
```

```python
# Load flight dataset
# Use on_bad_lines='skip' to handle rows with inconsistent column counts
# Use low_memory=False to resolve DtypeWarning from mixed types
df = pd.read_csv("Jan_2019_ontime.csv", on_bad_lines='skip', low_memory=False)

# Keep only required columns
df = df[['ORIGIN', 'DEST', 'DISTANCE']]

# Remove missing values
df = df.dropna()

# Remove duplicate routes
df = df.drop_duplicates()
```

```python
G = nx.Graph()

for _, row in df.iterrows():
    origin = row['ORIGIN']
    dest = row['DEST']
    distance = row['DISTANCE']

    G.add_edge(origin, dest, weight=distance)
```

```python
def uniform_cost_search(graph, start, goal):

    pq = [(0, start, [])]
    visited = set()

    while pq:

        cost, node, path = heapq.heappop(pq)

        if node in visited:
            continue

        visited.add(node)
        path = path + [node]

        if node == goal:
            return path, cost

        for neighbor in graph.neighbors(node):

            if neighbor not in visited:

                edge_cost = graph[node][neighbor]['weight']

                heapq.heappush(
                    pq,
                    (cost + edge_cost, neighbor, path)
                )

    return None, float("inf")
```

```python
def a_star(graph, start, goal):

    open_set = []

    heapq.heappush(open_set, (0, start))

    came_from = {}

    g_score = {start: 0}

    while open_set:

        _, current = heapq.heappop(open_set)

        if current == goal:

            path = [current]

            while current in came_from:
                current = came_from[current]
                path.append(current)

            path.reverse()

            return path, g_score[goal]

        for neighbor in graph.neighbors(current):

            tentative = g_score[current] + graph[current][neighbor]['weight']

            if neighbor not in g_score or tentative < g_score[neighbor]:

                came_from[neighbor] = current
                g_score[neighbor] = tentative

                f_score = tentative + heuristic(neighbor, goal)

                heapq.heappush(
                    open_set,
                    (f_score, neighbor)
                )

    return None, float("inf")
```

```python
start = "ATL"
goal = "LAX"
```

```python
start_time = time.time()

ucs_path, ucs_cost = uniform_cost_search(G, start, goal)

ucs_time = time.time() - start_time

print("Uniform Cost Search")

print("Path:", ucs_path)

print("Cost:", ucs_cost)

print("Execution Time:", ucs_time)
```

```python
start_time = time.time()

astar_path, astar_cost = a_star(G, start, goal)

astar_time = time.time() - start_time

print("\nA* Search")

print("Path:", astar_path)

print("Cost:", astar_cost)

print("Execution Time:", astar_time)
```
